```python
#  structuring right Binary encoding order for binary columns to ensure consistency and interpretability across the dataset
data["DifficultyWalking"] = data["DifficultyWalking"].replace({1:1,2:0})
data["Sex"] = data["Sex"].replace({1:1,2:0})
data["HeavyDrinking"] = data["HeavyDrinking"].replace({1:0,2:1})
data["HealthCareAccess"] = data["HealthCareAccess"].replace({1:1,2:0})
data["MedicalCost"] = data["MedicalCost"].replace({1:1,2:0})
data["PhysicalActivity"] = data["PhysicalActivity"].replace({1:1,2:0})
data["Fruits"] = data["Fruits"].replace({1:1,2:0})
data["Vegetables"] = data["Vegetables"].replace({1:1,2:0})
data["Smoke100"] = data["Smoke100"].replace({1:1,2:0})
data["Stroke"] = data["Stroke"].replace({1:1,2:0})
data["HeartDisease"] = data["HeartDisease"].replace({1:1,2:0})
data["HighBp"] = data["HighBp"].replace({1:0,2:1})
data["HighCholesterol"] = data["HighCholesterol"].replace({2:0,1:1})
data["CheckedCholesterol"] = data["CheckedCholesterol"].replace({1:1,2:0})


# Now 0 represents the absence of a risk factor and 1 represents the presence of a risk factor across all binary columns.
```

### Understanding missingness

```python
# Percentage of missing values in each column
missing_counts = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    "Missing Count": missing_counts,
    "Missing %": missing_percent
}).sort_values(by="Missing %", ascending=False)

print(missing_df)
```

```python
# Droping rows that have missing values greater than or equal to 50% of the columns.
# they are just too incomplete to be useful and could introduce noise into the model.
threshold = len(df.columns) * 0.5
df["missing_row_count"] = df.isnull().sum(axis=1)
df = df[df["missing_row_count"] < threshold]
df.drop(columns=["missing_row_count"], inplace=True)
```

```python
# Visualizing missingvalues of columns with missing percentage greater than 5% to understand the distribution of missing data and its relationship with diabetes status.

plt.figure(figsize=(20, 25)) 

for i, column in enumerate(missing_df[missing_df["Missing %"] > 5].index):
    plt.subplot(5, 5, i + 1) 
    sns.countplot(x=df[column].isnull(), hue=df['Diabetes'])
    plt.title(column)
    plt.xticks([0, 1], ['Present', 'Missing']) 
plt.tight_layout()
plt.show()
```

these above missingness may be informative since there are high gaps for their missing values by diabetes 

```python
plt.figure(figsize=(20,15))
sns.heatmap(df.isnull(), cbar=False, cmap='viridis')
plt.show()
```

**I'll classify the columns by their missingness based on the informations derived above** into "randomly_missing" and "structurally_missing" 
- **structurally_missing** (MAR):  will inpute but add flag 
- **randomly_missing** (MCAR): will only input 

```python
structurally_missing = ["HighCholesterol","Income","BMI","Fruits",
                        "Vegetables","HeavyDrinking","PhysicalActivity"]
# everything else not structurally missing here is randomly missing
```

```python
# creating new binary columns to indicate whether the value is missing or not for structurally missing columns.
for col in structurally_missing:
    df[col +"Missing"] = df[col].isnull().astype(int)
df.columns
```

```python
# applying appropriate imputation strategies for each type of missing data
# CONTINUOUS 
for col in continuous_cols:
    df[col] = df[col].fillna(df[col].median())

# BINARY
for col in binary_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

# ORDINAL
for col in ordinal_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

## Feature selection 

```python
#  copy of data to work with for feature selection 
Feature = data.copy()
Feature.columns
```

```python
# importing necessary libraries for feature selection.
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.compose import ColumnTransformer 
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder,RobustScaler
from sklearn.pipeline import Pipeline
```

```python
# missing flag for our structurally missing data   
for col in structurally_missing:
    Feature[f"{col}Missing"] = Feature[col].isnull().astype(int)
Feature.columns
```

```python
# defining features and target variables
X = Feature.drop(columns=["Diabetes"])
y = Feature["Diabetes"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1, stratify=y)
```

```python
# Continuous pipeline
continuous_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", RobustScaler())
])

# Binary and Ordinal  pipeline 
binord_pipeline = Pipeline(steps=[
    ("imputer",SimpleImputer(strategy="most_frequent"))
])

# Nominal Pipeline 
nominal_pipeline = Pipeline(steps=[
    ("Encoder",OneHotEncoder(drop="first", handle_unknown="ignore",sparse_output=False))
])

preprocessor = ColumnTransformer(transformers=[
    ("continuous",continuous_pipeline,continuous_cols),
    ("binary",binord_pipeline,binary_cols),
    ("ordinal",binord_pipeline,ordinal_cols),
    ("nominal",nominal_pipeline,nominal_cols)
],remainder="passthrough")
```

```python
# Transforming the data
X_train_clean = preprocessor.fit_transform(X_train)
X_test_clean = preprocessor.transform(X_test)

# Get feature names
feature_names = preprocessor.get_feature_names_out()

# train
RF_selector = RandomForestClassifier(random_state=1)
RF_selector.fit(X_train_clean, y_train)
```

```python
# renaming the feature names from the preprocessor  properly
feature_list = []
for names in feature_names:
    before,delimeter,after = names.partition("__")
    feature_list.append(after)
features = np.array(feature_list)
features
```

```python
#  DataFrame of features sorted by importance
importance_df = pd.DataFrame({
    "Feature":features,
    "Importance": RF_selector.feature_importances_
},).sort_values(by="Importance", ascending=False).reset_index(drop=True)

# Normalized importance 
importance_df["Normalized Importance"] = (importance_df["Importance"] - importance_df["Importance"].min()) / (importance_df["Importance"].max() - importance_df["Importance"].min())

# cummulative importance
importance_df["Cummulative Importance"] = importance_df["Normalized Importance"].cumsum()

# treshold for selecting features (i choose features within 85% of cummulative predicting power )
treshold = 0.85

#  Filter rows. We shift the threshold boundary by one to include the final boundary feature that pushes the cumulative sum over 85%.
selected_df = df[df["Cummulative Importance"].shift(1, fill_value=0) < threshold]
selected_features = selected_df["Feature"].to_numpy()


```

```python
# Creating a seaborn bar plot

fig, ax = plt.subplots(figsize=(30, 24))
ax = sns.barplot(x=importance_df["Importance"], y=importance_df["Features"], data=importance_df)
ax.set_title("Visualize feature scores of the features")
ax.set_yticklabels(importance_df["Features"])
ax.set_xlabel("Feature importance score")
ax.set_ylabel("Features")
plt.show()
```
